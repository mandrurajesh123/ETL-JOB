```python
import time

# Simulate streaming new data every few seconds
streaming_data = [
    ("David", 28),
    ("Eva", 32),
    ("Frank", 40),

    ("David", 28),
    ("Eva", 32),
    ("Frank", 40),

    ("David", 28),
    ("Eva", 32),
    ("Frank", 40),

    ("David", 28),
    ("Eva", 32),
    ("Frank", 40)
]

for record in streaming_data:

    new_df = spark.createDataFrame([record], columns)

    new_df.write \
        .format("delta") \
        .mode("append") \
        .save("/Volumes/my_bronze/bronze/bronze_volue/DSA_Delta_Table_Source4")

    print(f"Appended record: {record}")

    time.sleep(5)  # Simulate delay between records


```
